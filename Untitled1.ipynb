```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Set visual style
sns.set_theme(style="whitegrid")

# Load dataset
file_path = 'Dataset for Data Analytics.xlsx'
df = pd.read_excel(file_path)

# Quick Inspection
print("Dataset Shape:", df.shape)
print("\nData Types & Info:")
print(df.info())

# Check missing values
print("\nMissing Values Count:")
print(df.isnull().sum())
```

```python
# Clean missing values in CouponCode
df['CouponCode'] = df['CouponCode'].fillna('NO_COUPON')

# Verify that there are no missing values remaining
print("Remaining Missing Values:")
print(df.isnull().sum())
```

```python
# --- Step 2: Descriptive Statistics ---
numerical_cols = ['Quantity', 'UnitPrice', 'ItemsInCart', 'TotalPrice']

print("--- Five-Number Summary & Key Statistics ---")
summary_table = df[numerical_cols].describe().T[['mean', 'std', 'min', '25%', '50%', '75%', 'max']]
summary_table.columns = ['Mean', 'Std Dev', 'Min', 'Q1 (25%)', 'Median (50%)', 'Q3 (75%)', 'Max']
display(summary_table)
```

```python
# --- Step 3: Outlier Detection & Distributions ---
fig, axes = plt.subplots(1, 2, figsize=(14, 4))

# Distribution plot for TotalPrice
sns.histplot(df['TotalPrice'], kde=True, color='teal', ax=axes[0])
axes[0].set_title('Distribution of Total Price')
axes[0].set_xlabel('Total Price ($)')

# Boxplot for TotalPrice
sns.boxplot(x=df['TotalPrice'], color='lightblue', ax=axes[1])
axes[1].set_title('Boxplot of Total Price (IQR Rule)')
axes[1].set_xlabel('Total Price ($)')

plt.tight_layout()
plt.show()

# Calculate Outliers using Interquartile Range (IQR)
Q1 = df['TotalPrice'].quantile(0.25)
Q3 = df['TotalPrice'].quantile(0.75)
IQR = Q3 - Q1
upper_bound = Q3 + 1.5 * IQR

outliers = df[df['TotalPrice'] > upper_bound]
print(f"IQR: {IQR:.2f}")
print(f"Upper Boundary for Outliers: ${upper_bound:.2f}")
print(f"Number of Outlier Orders Found: {len(outliers)}")
```

```python
# --- Step 4: Product Revenue & Correlation Matrix ---
plt.figure(figsize=(10, 4))

# Product Revenue Ranking
product_sales = df.groupby('Product')['TotalPrice'].sum().reset_index().sort_values(by='TotalPrice', ascending=False)
sns.barplot(data=product_sales, x='TotalPrice', y='Product', palette='viridis')
plt.title('Total Revenue Generated per Product')
plt.xlabel('Total Revenue ($)')
plt.ylabel('Product Category')
plt.show()

# Correlation Matrix
plt.figure(figsize=(6, 4))
sns.heatmap(df[numerical_cols].corr(), annot=True, cmap='coolwarm', fmt=".2f")
plt.title('Pearson Correlation Matrix')
plt.show()
```

```python
# --- Pivot Table: Total Sales by Product and Payment Method ---
pivot_summary = pd.pivot_table(
    df,
    values='TotalPrice',
    index='Product',
    columns='PaymentMethod',
    aggfunc='sum'
)

print("--- Sales Pivot Table (Product vs Payment Method) ---")
display(pivot_summary)
```

# Executive Summary & Business Insights

### 1. Problem Statement
Analyze 1,200 transaction records to identify sales patterns, outlier behaviors, product performance, and operational optimizations for revenue growth.

### 2. Data Forensics & Methodology
- **Dataset Scope:** 1,200 orders across 14 transactional metrics.
- **Data Cleaning:** Imputed 309 missing values in `CouponCode` with `"NO_COUPON"`, achieving 100% data completeness across all columns.
- **Outlier Diagnostic:** Calculated IQR boundaries and identified 8 upper-tail outlier orders exceeding $3,330.41.

### 3. Key Findings
- **Revenue Distribution:** Mean order value ($1,053.97) exceeds the median ($823.62), indicating a right-skewed revenue distribution.
- **Correlation Drivers:** `TotalPrice` shows strong positive correlations with `Quantity` (0.62) and `UnitPrice` (0.65), demonstrating that bulk purchases and high-ticket hardware drive total order volume.
- **Product Rankings:** Chairs ($195,620) and Printers ($195,612) are the top revenue generators, while Phones ($151,722) yield the lowest total sales.

### 4. Recommendations
- **Promote Bundled Deals:** Create incentives around high-ticket items (Printers, Laptops) to capitalize on the strong correlation between item price, quantity, and overall order value.
- **Track Promotion Effectiveness:** Monitor the 25.75% non-coupon orders against coupon-bearing orders to refine discount strategies and protect profit margins.
